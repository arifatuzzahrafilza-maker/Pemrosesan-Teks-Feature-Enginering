**One Hot Encoding**

Teknik yang digunakan dalam machine learning dan pengolahan data untuk mengubah variabel kategorikal (data berupa teks atau kategori) menjadi format angka biner (0 dan 1) agar dapat dipahami oleh algoritma komputer.

Proses pelaksanaan One-Hot Encoding umumnya meliputi:

1. Mengidentifikasi Kategori Unik: Memindai kolom data kategorikal untuk menemukan seluruh nilai unik (unique values) atau kategori yang ada.

2. Membuat Kolom Biner Baru: Membuat kolom baru sejumlah kategori unik yang telah ditemukan sebelumnya.

3. Mengisi Nilai Biner: Memberikan nilai 1 pada kolom yang sesuai dengan kategori asli baris data tersebut, dan mengisi nilai 0 untuk semua kolom kategori lainnya.

```python
import re
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

# 1. Baca file CSV
df = pd.read_csv('scrapinx_preprocessed (1).csv')

# 2. Hapus data duplikat pada kolom final_text
df = df.drop_duplicates(subset=['final_text']).reset_index(drop=True)

# 3. Fungsi pembersihan tambahan
def bersihkan_teks(teks):
    teks = str(teks)
    # Ubah kata pengulangan berangka (misal: orang2 -> orang)
    teks = re.sub(r'(\w+)2\b', r'\1', teks)
    # Hapus semua angka
    teks = re.sub(r'\d+', '', teks)
    # Hapus karakter non-alfabet (simbol/tanda baca)
    teks = re.sub(r'[^a-zA-Z\s]', '', teks)
    # Hapus spasi ganda
    teks = re.sub(r'\s+', ' ', teks).strip()
    return teks

# Terapkan pembersihan pada data
df['clean_final'] = df['final_text'].apply(bersihkan_teks)

# 4. Melakukan One-Hot Encoding
# min_df=2 digunakan untuk mengabaikan kata typo/aneh yang cuma muncul 1x di seluruh dataset
vectorizer = CountVectorizer(binary=True, min_df=2)
X = vectorizer.fit_transform(df['clean_final'])

# 5. Buat DataFrame hasil One-Hot Encoding
df_onehot = pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())
df_onehot
```

**TF-IDF**

Metode statistik yang digunakan untuk mengukur seberapa penting atau relevan suatu kata terhadap sebuah dokumen di dalam kumpulan dokumen (korpus). Metode ini membantu mengubah teks menjadi angka dengan memberikan bobot tinggi pada kata yang sering muncul di satu dokumen tertentu tetapi jarang muncul di dokumen lainnya.

Proses perhitungan TF-IDF umumnya melalui tahapan-tahapan berikut:

1. Menghitung Term Frequency (TF):
Menghitung frekuensi atau seberapa sering sebuah kata muncul di dalam suatu dokumen tertentu. Semakin sering kata tersebut muncul, semakin tinggi nilai TF-nya.

2. Menghitung Inverse Document Frequency (IDF):
Mengukur tingkat kelangkaan atau kepentingan suatu kata di seluruh kumpulan dokumen. Kata-kata yang terlalu umum (seperti "dan", "yang", "di") akan diberikan bobot IDF yang kecil, sedangkan kata yang spesifik dan unik akan mendapatkan bobot IDF yang besar.

3. Mengalikan TF dan IDF:
Mengalikan skor TF dengan skor IDF untuk setiap kata. Hasil perkalian inilah yang menjadi bobot akhir TF-IDF, di mana kata yang sering muncul dalam satu dokumen tetapi jarang di dokumen lain akan memiliki nilai bobot yang paling tinggi.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# 1. Melakukan TF-IDF Vectorization pada teks yang sudah bersih
# min_df=2 digunakan agar konsisten mengabaikan kata typo/aneh yang cuma muncul 1x
vectorizer_tfidf = TfidfVectorizer(min_df=2)
X_tfidf = vectorizer_tfidf.fit_transform(df['clean_final'])

# 2. Buat DataFrame hasil TF-IDF
df_tfidf = pd.DataFrame(X_tfidf.toarray(), columns=vectorizer_tfidf.get_feature_names_out())
df_tfidf
```

**BUILD WORD2VEC MODEL**



**Word2vec Model**

Teknik word embedding berbasis neural network yang dikembangkan oleh Google untuk mengubah kata-kata menjadi representasi vektor numerik. Tujuannya adalah agar kata-kata yang memiliki arti atau konteks serupa ditempatkan berdekatan dalam ruang vektor, sehingga komputer dapat memahami hubungan semantik antar kata.

Proses pembuatan dan pelatihan model Word2Vec umumnya melalui tahapan-tahapan berikut:

1. Tokenisasi dan Pembersihan Teks: Memecah teks menjadi satuan kata individual (tokens) dan melakukan penyesuaian seperti mengubah huruf menjadi kecil (lowercase) serta membersihkan karakter yang tidak diperlukan.

2. Pembangunan Kamus (Vocabulary Building): Mengumpulkan seluruh kata unik yang ada di dalam korpus teks dan memetakannya ke dalam indeks numerik.

3. Pembuatan Pasangan Konteks dan Target: Menggunakan jendela geser (sliding window) untuk menentukan kata target dan kata-kata di sekitarnya berdasarkan arsitektur model (seperti CBOW untuk memprediksi kata dari konteks, atau Skip-gram untuk memprediksi konteks dari kata).

4. Inisialisasi Vektor Acak: Memberikan nilai vektor numerik awal secara acak untuk setiap kata di dalam kamus.

5. Pelatihan Jaringan Saraf Tiruan (Neural Network Training): Melatih model secara iteratif untuk memprediksi hubungan kata, sehingga bobot vektor pada setiap kata terus diperbarui sampai merepresentasikan kedekatan makna yang akurat.

```python
# 5. FE result
import numpy as np
def get_document_vector(tokens, model):
    valid_vectors = [model.wv[word] for word in tokens if word in model.wv]
    # if not valid_vectors:
    #   return np.zeros(model.vector_size)
    return np.mean(valid_vectors, axis=0)
X_features = np.array([get_document_vector(doc, model) for doc in sentences])
pd.DataFrame(X_features)
```

**BUILD FAST TEXT MODEL**

**FastText**

Teknik word embedding tingkat lanjut yang dikembangkan oleh Facebook (Meta) sebagai pengembangan dari Word2Vec. Jika Word2Vec melihat setiap kata sebagai satu kesatuan utuh, FastText memperlakukan setiap kata sebagai gabungan dari potongan-potongan karakter yang lebih kecil atau disebut sub-kata (subwords / n-grams). Hal ini membuat FastText jauh lebih handal dalam mengenali kata-kata baru, singkatan gaul, maupun kata yang mengalami salah ketik (typo) yang sering ditemukan pada data teks media sosial seperti tweet.

Proses pelatihan dan pembentukan model FastText umumnya melalui tahapan-tahapan berikut:

1. Tokenisasi dan Pembersihan Teks:
Memecah dokumen atau tweet menjadi daftar kata individual (tokens) serta melakukan penyesuaian dasar seperti mengubah huruf menjadi kecil (lowercase).

2. Ekstraksi Sub-Kata (Subword N-grams Extraction):
Memecah setiap kata menjadi bagian-bagian karakter yang lebih kecil berdasarkan panjang tertentu (misalnya n-gram dari 3 hingga 6 karakter). Sebagai contoh, kata "makan" akan dipecah menjadi potongan seperti <ma, mak, akan, kan>, termasuk simbol penanda awal dan akhir kata.

3. Pembangunan Kamus dan Vektor Sub-Kata:
Mengumpulkan seluruh kata unik serta kamus dari seluruh sub-kata yang terbentuk, lalu menginisialisasi vektor numerik awal untuk masing-masing sub-kata tersebut.

4. Pembentukan Vektor Kata Utuh:
Merepresentasikan vektor dari sebuah kata sebagai hasil penjumlahan atau gabungan dari vektor seluruh sub-kata yang membentuk kata tersebut (ditambah dengan representasi kata utuhnya).

5. Pelatihan Model (Training):
Melatih jaringan saraf tiruan secara iteratif untuk mempelajari konteks kemunculan kata. Melalui proses ini, bobot vektor pada setiap sub-kata dan kata terus diperbarui agar model mampu memahami hubungan makna serta struktur ejaan kata dengan akurat.

**Perbedaan Visualization Word2vec dan FastText**

Perbedaan utama antara visualisasi FastText dan Word2Vec (yang dipetakan melalui UMAP) terletak pada bagaimana kedua model mengelompokkan kata-kata yang mengalami salah ketik (typo), singkatan gaul, serta bentuk kata berimbuhan.



*   Word2Vec: Setiap kata diperlakukan sebagai entitas utuh, sehingga kata baru atau istilah slang yang tidak terdapat dalam kamus pelatihan (Out-of-Vocabulary) sama sekali tidak dapat dimunculkan atau dipetakan dalam grafik. Selain itu, sebaran titik kata pada data media sosial seperti tweet cenderung terisolasi dan kurang rapat karena hubungan antar-kata murni didasarkan pada konteks kemunculan, bukan struktur ejaannya.
*   FastText: Memanfaatkan informasi sub-kata (n-grams), sehingga mampu menampilkan kata-kata baru, singkatan khas Twitter, maupun typo dengan menempatkannya dalam klaster yang sangat berdekatan dengan kata dasarnya. Hal ini menghasilkan sebaran grafik UMAP yang lebih padat, terstruktur, dan merepresentasikan kemiripan ejaan sekaligus makna secara lebih intuitif.
